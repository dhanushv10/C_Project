```python
#line plot
plt.figure(figsize=(5, 3))
plt.plot(x, y, marker="o", color="b", linestyle="--")
plt.title("Line Plot Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(5, 3))
plt.bar(categories, values, color="skyblue")
plt.title("Bar Plot Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(5, 3))
plt.scatter(x, y, color="red", s=100)  # s controls point size
plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt


data = {
    "Hours_Studied": [2, 5, 7, 8, 10],
    "Attendance_%": [60, 75, 80, 90, 95],
    "Exam_Score": [50, 65, 78, 88, 98]
}
df = pd.DataFrame(data)

matrix = df.corr()

plt.figure(figsize=(5, 4))
sns.heatmap(matrix, annot=True, cmap="YlGnBu")
plt.title("Study Habits vs Score Correlation")
plt.show()


sns.pairplot(df)
plt.show()
```
